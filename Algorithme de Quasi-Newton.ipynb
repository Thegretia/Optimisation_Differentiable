```python

import numpy as np

def quasi_newton(f, grad, x0, tol=1e-6, max_iter=1000):

    x = np.array(x0, dtype=float)
    n = len(x)

    H = np.eye(n)

    for k in range(max_iter):


        g = grad(x)

        if np.linalg.norm(g) < tol:
            break

        d = -H @ g

    
        alpha = 1.0
        rho = 0.5
        c = 1e-4

        while f(x + alpha * d) > f(x) + c * alpha * np.dot(g, d):
            alpha *= rho

            if alpha < 1e-12:
                break


        x_new = x + alpha * d


        g_new = grad(x_new)


        s = x_new - x
        y = g_new - g

        ys = np.dot(y, s)

        if ys > 1e-12:
            rho_bfgs = 1.0 / ys

            I = np.eye(n)

            H = (
                (I - rho_bfgs * np.outer(s, y))
                @ H
                @ (I - rho_bfgs * np.outer(y, s))
                + rho_bfgs * np.outer(s, s)
            )

    
        x = x_new

    return x, f(x), k + 1
```

```python

def f(x):
    return (x[0] - 2)**2 + (x[1] - 3)**2


def grad(x):
    return np.array([
        2 * (x[0] - 2),
        2 * (x[1] - 3)
    ])

x0 = [0, 0]

solution, valeur_min, iterations = quasi_newton(
    f, grad, x0
)

print("Solution optimale :", solution)
print("Valeur minimale :", valeur_min)
print("Nombre d'itérations :", iterations)
```
