```python

import numpy as np

# Fonction à minimiser
def f(X):
    x, y, z = X
    return (1/4)*z**4 + y**3 - (1/3)*z**3 + 6*x**2 + 3*y**2 + 6*x*y


# Gradient de la fonction
def gradient(X):
    x, y, z = X
    return np.array([
        12*x+6*y,
        3*y**2 + 6*y + 6*x,
        3*z**3-z**2
    ])


# Matrice hessienne
def hessienne(X):
    x, y, z = X
    return np.array([
        [ 12, 6, 0],
        [6, 6*y + 6, 0],
        [0, 0, 3*z**2-2*z]
    ])


# Algorithme de Newton
def newton(X0, tol=1e-5, max_iter=100):
    X = np.array(X0, dtype=float)

    print("Itération 0 :", X, " f(X) =", f(X))

    for k in range(max_iter):
        G = gradient(X)
        H = hessienne(X)

        # Critère d'arrêt
        if np.linalg.norm(G) < tol:
            print("Convergence atteinte.")
            break

        try:
            D = np.linalg.solve(H, G)
        except np.linalg.LinAlgError:
            print("La matrice hessienne est singulière.")
            break

        X = X - D

        print(f"Itération {k+1} : X = {X}, f(X) = {f(X):.8f}")

    print("\nRésultat final :")
    print("X =", X)
    print("f(X) =", f(X))
    print("Norme du gradient =", np.linalg.norm(gradient(X)))

    return X


# Point initial demandé
X0 = [-0.667, 0.1728, 1.5778]

# Exécution
solution = newton(X0)
```
