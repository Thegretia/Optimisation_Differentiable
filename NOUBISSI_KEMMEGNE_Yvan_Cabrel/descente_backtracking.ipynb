```python
# Les fonctions communes sont dans fonctions.py.
import fonctions as fct
```

```python
def descente_backtracking(f, x0, grad=None, eps=1e-6, max_iter=10000,
                          alpha0=1.0, beta=0.5, c=1e-4):
    """Descend le gradient et reduit le pas jusqu'e obtenir une baisse de f."""
    # Si le gradient n'est pas fourni, on l'estime a partir de f.
    if grad is None:
        grad = lambda x: fct.gradient_numerique(f, x)

    x = x0[:]  # On garde x0 intact.
    historique = []

    for k in range(max_iter + 1):
        gradient = grad(x)
        longueur_gradient = fct.norme(gradient)
        historique.append((k, f(x), longueur_gradient))

        if longueur_gradient <= eps or k == max_iter:
            break

        # Le pas de backtracking est reduit automatiquement si necessaire.
        direction = fct.produit_par_scalaire(-1.0, gradient)
        pas = fct.backtracking(f, x, gradient, direction, alpha0, beta, c)
        deplacement = fct.produit_par_scalaire(pas, direction)
        x = fct.somme(x, deplacement)

    return x, historique
```

```python
EPS = 1e-6

# Le pas de depart vaut 1. Le backtracking le reduit si le deplacement est trop grand.
x, historique = descente_backtracking(
    fct.f_quadratique, fct.X0_QUAD,
    grad=fct.grad_quadratique, eps=EPS
)
fct.afficher_historique(historique)
print()
fct.resume("Backtracking - quadratique", x, historique, fct.X_EXACT_QUAD, EPS)
```

```python
# Meme methode sur Rosenbrock, qui demande davantage d'iterations.
x, historique = descente_backtracking(
    fct.f_rosenbrock, fct.X0_ROSEN,
    grad=fct.grad_rosenbrock, eps=EPS, max_iter=50000
)
fct.afficher_historique(historique)
print()
fct.resume("Backtracking - Rosenbrock", x, historique, fct.X_EXACT_ROSEN, EPS)
```

```python
# Sans argument grad, l'algorithme estime le gradient a partir de f.
x, historique = descente_backtracking(
    fct.f_quadratique, fct.X0_QUAD, grad=None, eps=EPS
)
fct.resume("Backtracking - gradient numerique", x, historique, fct.X_EXACT_QUAD, EPS)
```
