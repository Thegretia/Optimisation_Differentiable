```python
# Les fonctions communes sont dans fonctions.py.
import fonctions as fct
```

```python
def descente_pas_fixe(f, x0, alpha, grad=None, eps=1e-6, max_iter=10000):
    """Minimise f avec un pas alpha qui reste le meme a chaque iteration."""
    # Si le gradient n'est pas fourni, on l'estime a partir de f.
    if grad is None:
        grad = lambda x: fct.gradient_numerique(f, x)

    x = x0[:]  # On garde x0 intact.
    historique = []

    for k in range(max_iter + 1):
        gradient = grad(x)
        longueur_gradient = fct.norme(gradient)
        historique.append((k, f(x), longueur_gradient))

        # Un petit gradient signifie que le minimum est suffisamment proche.
        if longueur_gradient <= eps:
            break
        if k == max_iter:
            break

        # On avance dans la direction opposee au gradient.
        deplacement = fct.produit_par_scalaire(-alpha, gradient)
        x = fct.somme(x, deplacement)

    return x, historique
```

```python
# Tolerance : l'algorithme s'arrete quand le gradient est assez petit.
EPS = 1e-6

# Pour cette quadratique, on choisit un pas fixe a partir des valeurs propres de A.
lambda_min, lambda_max = fct.valeurs_propres_symetrique(fct.A_QUAD)
kappa = lambda_max / lambda_min
alpha = 2 / (lambda_min + lambda_max)

print(f"Plus petite valeur propre : {lambda_min:.4f}")
print(f"Plus grande valeur propre : {lambda_max:.4f}")
print(f"Pas fixe choisi : {alpha:.5f}")

# On lance la descente avec le gradient connu de la quadratique.
x, historique = descente_pas_fixe(
    fct.f_quadratique, fct.X0_QUAD, alpha,
    grad=fct.grad_quadratique, eps=EPS
)
fct.afficher_historique(historique)
print()
fct.resume("Pas fixe - quadratique", x, historique, fct.X_EXACT_QUAD, EPS)
```

```python
# Rosenbrock a un minimum en [1, 1, 1].
# On prend un pas petit pour eviter que les iterations divergent.
x, historique = descente_pas_fixe(
    fct.f_rosenbrock, fct.X0_ROSEN, alpha=1e-3,
    grad=fct.grad_rosenbrock, eps=EPS, max_iter=50000
)
fct.afficher_historique(historique)
print()
fct.resume("Pas fixe - Rosenbrock", x, historique, fct.X_EXACT_ROSEN, EPS)
```
