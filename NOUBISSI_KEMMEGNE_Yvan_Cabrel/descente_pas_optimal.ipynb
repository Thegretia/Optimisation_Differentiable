```python
# Les fonctions communes sont dans fonctions.py.
import fonctions as fct
```

```python
def descente_pas_optimal(f, x0, grad=None, eps=1e-6, max_iter=10000):
    """Minimise f en choisissant le meilleur pas a chaque iteration."""
    # Si le gradient n'est pas fourni, on l'estime a partir de f.
    if grad is None:
        grad = lambda x: fct.gradient_numerique(f, x)

    x = x0[:]  # On garde x0 intact.
    historique = []

    for k in range(max_iter + 1):
        gradient = grad(x)
        longueur_gradient = fct.norme(gradient)
        historique.append((k, f(x), longueur_gradient))

        if longueur_gradient <= eps or k == max_iter:
            break

        # La direction de descente est l'oppose du gradient.
        direction = fct.produit_par_scalaire(-1.0, gradient)
        pas = fct.pas_optimal(f, x, direction)
        deplacement = fct.produit_par_scalaire(pas, direction)
        x = fct.somme(x, deplacement)

    return x, historique
```

```python
EPS = 1e-6

# Verifie le pas calcule par recherche doree sur une fonction quadratique.
gradient_initial = fct.grad_quadratique(fct.X0_QUAD)
direction = fct.produit_par_scalaire(-1.0, gradient_initial)
pas_numerique = fct.pas_optimal(fct.f_quadratique, fct.X0_QUAD, direction)
pas_exact = fct.pas_optimal_quadratique(fct.A_QUAD, gradient_initial)

print(f"Pas par recherche doree : {pas_numerique:.10f}")
print(f"Pas exact              : {pas_exact:.10f}")
print(f"Difference             : {abs(pas_numerique - pas_exact):.2e}\n")

# Lance la descente sur la quadratique avec son gradient exact.
x, historique = descente_pas_optimal(
    fct.f_quadratique, fct.X0_QUAD,
    grad=fct.grad_quadratique, eps=EPS
)
fct.afficher_historique(historique)
print()
fct.resume("Pas optimal - quadratique", x, historique, fct.X_EXACT_QUAD, EPS)
```

```python
# La recherche doree choisit le pas automatiquement a chaque iteration.
x, historique = descente_pas_optimal(
    fct.f_rosenbrock, fct.X0_ROSEN,
    grad=fct.grad_rosenbrock, eps=EPS, max_iter=20000
)
fct.afficher_historique(historique)
print()
fct.resume("Pas optimal - Rosenbrock", x, historique, fct.X_EXACT_ROSEN, EPS)
```

```python
# Sans argument grad, l'algorithme estime le gradient a partir de f.
x, historique = descente_pas_optimal(
    fct.f_quadratique, fct.X0_QUAD, grad=None, eps=EPS
)
fct.resume("Pas optimal - gradient numerique", x, historique, fct.X_EXACT_QUAD, EPS)
```
