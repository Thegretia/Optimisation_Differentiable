# Gradient à pas fixe

**Principe** : on avance dans la direction opposée au gradient avec un pas constant α :

x_{k+1} = x_k − α ∇f(x_k)

**Problème test** : f(x) = ½ xᵀAx − bᵀx avec A = [[6, 2], [2, 3]], b = [2, 1], x₀ = [3, −2].
Le gradient est ∇f(x) = Ax − b. Le minimiseur exact est x* = [2/7, 1/7].

**Exemple à la main (α = 0,1)** :
∇f(x₀) = Ax₀ − b = [12, −1], donc x₁ = [3, −2] − 0,1·[12, −1] = [1,8 ; −1,9].

**Condition de convergence** : le pas fixe converge si 0 < α < 2/λmax = 2/7 ≈ 0,286.

```python
import numpy as np

A = np.array([[6.0, 2.0], [2.0, 3.0]])
b = np.array([2.0, 1.0])

f = lambda x: 0.5 * x @ A @ x - b @ x
grad_f = lambda x: A @ x - b

def gradient_pas_fixe(grad, x0, alpha, tol=1e-6, max_iter=10000):
    x = np.array(x0, dtype=float)
    xs = [x.copy()]                      # historique des itérés
    normes = [np.linalg.norm(grad(x))]   # historique de ||gradient||
    for k in range(max_iter):
        g = grad(x)
        if np.linalg.norm(g) < tol:      # critère d'arrêt
            break
        x = x - alpha * g                # x_{k+1} = x_k - alpha * grad
        xs.append(x.copy())
        normes.append(np.linalg.norm(grad(x)))
        if normes[-1] > 1e100:           # garde-fou : divergence
            print("Divergence à l'itération", k + 1)
            break
    return x, xs, normes

x0 = [3.0, -2.0]
x_star = np.linalg.solve(A, b)

# Test avec alpha = 0.1 (valide)
x, xs, normes = gradient_pas_fixe(grad_f, x0, 0.1)
print("x =", x, "| itérations =", len(xs) - 1)
print("erreur =", np.linalg.norm(x - x_star))
print("x1 =", xs[1])                     # doit valoir [1.8, -1.9]

# Test avec alpha = 0.3 (trop grand : divergence)
gradient_pas_fixe(grad_f, x0, 0.3, max_iter=100)
```

```python
x0 = [3.0, -2.0]
x_star = np.linalg.solve(A, b)

x, xs, normes = gradient_pas_fixe(grad_f, x0, 0.1)
print("x =", x, "| itérations =", len(xs) - 1)
print("erreur =", np.linalg.norm(x - x_star))
print("x1 =", xs[1])
```

**Conclusion** : avec α = 0,1, l'algorithme converge en 71 itérations. Avec α = 0,3 (trop grand), il diverge.
