# Algorithme de Newton

**Principe** : on remplace f par son développement de Taylor d'ordre 2 en x_k, puis on saute au minimum de ce modèle. La direction d_k résout :

∇²f(x_k) d_k = −∇f(x_k), puis x_{k+1} = x_k + d_k (pas égal à 1)

On résout un système linéaire plutôt que d'inverser la hessienne : c'est plus stable.

**Problème test** : f(x) = ½ xᵀAx − bᵀx avec A = [[6, 2], [2, 3]], b = [2, 1], x₀ = [3, −2].
Ici ∇f(x) = Ax − b et la hessienne est constante, égale à A.

**Exemple à la main** :
- g₀ = [12, −1]
- A⁻¹ = (1/14)·[[3, −2], [−2, 6]], donc A⁻¹ g₀ = [19/7, −15/7]
- x₁ = [3 − 19/7, −2 + 15/7] = [2/7, 1/7] = x*

```python
import numpy as np

A = np.array([[6.0, 2.0], [2.0, 3.0]])
b = np.array([2.0, 1.0])
f = lambda x: 0.5 * x @ A @ x - b @ x
grad_f = lambda x: A @ x - b
hess_f = lambda x: A

def newton(grad, hess, x0, tol=1e-6, max_iter=100):
    x = np.array(x0, dtype=float)
    xs = [x.copy()]
    normes = [np.linalg.norm(grad(x))]
    for k in range(max_iter):
        g = grad(x)
        if np.linalg.norm(g) < tol:
            break
        d = np.linalg.solve(hess(x), -g)   # on résout H d = -g
        x = x + d                          # pas égal à 1
        xs.append(x.copy())
        normes.append(np.linalg.norm(grad(x)))
    return x, xs, normes

x0 = [3.0, -2.0]
x_star = np.linalg.solve(A, b)
x, xs, normes = newton(grad_f, hess_f, x0)

print("x =", x, "| itérations =", len(xs) - 1)
print("erreur =", np.linalg.norm(x - x_star))
print("x1 =", xs[1])
```

**Conclusion** : Newton converge en 1 seule itération, avec une erreur de l'ordre de 10⁻¹⁶. C'est normal : f est exactement quadratique, donc son modèle de Taylor d'ordre 2 est f elle-même. En contrepartie, Newton exige de calculer la hessienne et de résoudre un système linéaire à chaque itération, ce qui devient coûteux en grande dimension.
