# Descente de gradient a pas optimal

L'objectif `f` est une expression symbolique en trois variables. L'exemple quadratique du code peut etre remplace par la fonction etudiee. Le pas calcule ci-dessous est exact pour une fonction quadratique de hessienne constante.

```python
import numpy as np
import sympy as sp
```

```python
x1, x2, x3 = sp.symbols("x1 x2 x3")
variables = (x1, x2, x3)

# Exemple quadratique convexe a remplacer par la fonction etudiee.
f = x1**2 + 2*x2**2 + 3*x3**2
gradient_symbolique = sp.Matrix([sp.diff(f, variable) for variable in variables])
hessienne_symbolique = sp.hessian(f, variables)

f_num = sp.lambdify(variables, f, "numpy")
gradient_num = sp.lambdify(variables, gradient_symbolique, "numpy")
hessienne_num = sp.lambdify(variables, hessienne_symbolique, "numpy")


def grad_f(point):
    return np.array(gradient_num(point[0], point[1], point[2]), dtype=float).reshape(3)


Q = np.array(hessienne_num(0, 0, 0), dtype=float)
```

```python
x_0 = np.array([4.0, -3.0, 2.0])
x_k = x_0.copy()
eps = 1e-6
k = 0
```

## Iteration

```python
gradient_k = grad_f(x_k)

while np.sqrt(gradient_k @ gradient_k) > eps :
    alpha_k = (gradient_k @ gradient_k) / (gradient_k @ Q @ gradient_k)
    x_k = x_k - alpha_k * gradient_k
    gradient_k = grad_f(x_k)
    k += 1

print("Nombre d'iterations :", k)
print("Point final :", x_k)
print("Valeur finale de f :", f_num(x_k[0], x_k[1], x_k[2]))
print("Norme du gradient final :", np.sqrt(gradient_k @ gradient_k))
```
